```python
import numpy as np
import cv2
import os
import matplotlib.pyplot as plt
from tensorflow.keras import layers, models
```

```python
# CLASS NAMES
classes = ["Cat", "Dog"]
```

```python
data_image="/content/sample_data/images"
```

```python
images = []
labels = []
```

```python
for i, file in enumerate(os.listdir(data_image)):

    # Complete image path
    path = os.path.join(data_image, file)

    # Read image
    img = cv2.imread(path)

    # Skip if image not loaded
    if img is None:
        continue

    # Resize image
    img = cv2.resize(img, (32, 32))

    # Normalize image
    img = img / 255.0

    # Store image
    images.append(img)

    # Dummy labels
    labels.append(i % 2)

# -----------------------------------------
# CONVERT TO NUMPY ARRAY
# -----------------------------------------

images = np.array(images)
labels = np.array(labels)

print("Images Shape :", images.shape)
print("Labels Shape :", labels.shape)

# -----------------------------------------
# DISPLAY SAMPLE IMAGE
# -----------------------------------------

plt.imshow(images[0])
plt.title(classes[labels[0]])
plt.show()
```

```python
# BUILD CNN MODEL
# -----------------------------------------

model = models.Sequential([

    # Input Layer
    layers.Input(shape=(64, 64, 3)),

    # Convolution Layer
    layers.Conv2D(32, (3,3), activation='relu'),

    # Pooling Layer
    layers.MaxPooling2D((2,2)),

    # Convert 2D to 1D
    layers.Flatten(),

    # Dense Layer
    layers.Dense(64, activation='relu'),

    # Output Layer
    layers.Dense(2, activation='softmax')
])
```

```python
# COMPILE MODEL

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# MODEL SUMMARY

model.summary()

# -----------------------------------------
# TRAIN MODEL
# -----------------------------------------
model.fit(
    images,
    labels,
    epochs=5
)

# -----------------------------------------
# PREDICT IMAGE
# -----------------------------------------

prediction = model.predict(np.array([images[0]]))

index = np.argmax(prediction)

print("\nPredicted Class :", classes[index])
```
